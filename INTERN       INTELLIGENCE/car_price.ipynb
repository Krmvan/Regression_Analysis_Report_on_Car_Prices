```python
import statsmodels.api as sm
```

```python
df=df[['Engine_Size','Mileage','Doors','Owner_Count','Price']]
```

```python
df.head()
```

```python
X=df.drop(['Price'],axis=1)
```

```python
X.head()
```

```python
X=X.assign(const=1)
```

```python
X.head()
```

```python
y=df['Price']
```

```python
model = sm.OLS(y,X)
results = model.fit()
```

```python
print(results.summary())
```

```python
import seaborn as sns
```

```python
import matplotlib.pyplot as plt
```

```python

# Engine Size and Price 
sns.regplot(x=df["Engine_Size"], y=df["Price"], scatter_kws={"alpha":0.5}, line_kws={"color":"red"})
plt.title("Engine Size vs Price")
plt.show()

# Mileage and Price
sns.regplot(x=df["Mileage"], y=df["Price"], scatter_kws={"alpha":0.5}, line_kws={"color":"red"})
plt.title("Mileage vs Price")
plt.show()
```

```python
import pandas as pd

coef_df = pd.DataFrame({"Feature": ["Engine_Size", "Mileage", "Doors", "Owner_Count"], 
                        "Coefficient": [990.9018, -0.0201, 23.8897, 7.8710]})

coef_df.plot(x="Feature", y="Coefficient", kind="bar", legend=False)
plt.title("Feature Importance in Car Price Prediction")
plt.xlabel("Features")
plt.ylabel("Coefficient Value")
plt.show()
```
